```python
import pandas as pd

df = pd.read_excel("CodeOrbit_Task1_Raw_Dataset_500x10.xlsx")

print("Dataset loaded successfully!")
```

```python
# STEP 2: DATA INSPECTION

print("===== DATASET SHAPE =====")
print("Rows and Columns:", df.shape)

print("\n===== COLUMN NAMES =====")
print(df.columns.tolist())

print("\n===== FIRST 5 ROWS =====")
display(df.head())

print("\n===== LAST 5 ROWS =====")
display(df.tail())

print("\n===== DATASET INFORMATION =====")
df.info()
```

```python
# STEP 3: CHECK DATA QUALITY

print("===== MISSING VALUES =====")
print(df.isnull().sum())

print("\n===== DUPLICATE ROWS =====")
print("Number of duplicate rows:", df.duplicated().sum())

print("\n===== UNIQUE VALUES IN EACH COLUMN =====")
for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique()[:20])
```

```python
# STEP 4: CHECK DATA TYPES AND VALUES

print("===== DATA TYPES =====")
print(df.dtypes)

print("\n===== QUANTITY VALUES =====")
print(df["Quantity"].unique())

print("\n===== SALES AMOUNT VALUES =====")
print(df["Sales_Amount"].unique()[:20])

print("\n===== ORDER DATE VALUES =====")
print(df["Order_Date"].head(10))
```

```python
# STEP 5: DATA CLEANING

# 1. Remove duplicate rows
df = df.drop_duplicates()

# 2. Remove extra spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

# 3. Standardize text capitalization
for column in text_columns:
    df[column] = df[column].str.title()

# 4. Convert Quantity to numeric
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

# 5. Convert Sales_Amount to numeric
df["Sales_Amount"] = pd.to_numeric(df["Sales_Amount"], errors="coerce")

# 6. Convert Order_Date to date format
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

# 7. Handle missing values
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")
df["Category"] = df["Category"].fillna("Unknown")
df["Product"] = df["Product"].fillna("Unknown")
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
df["Order_Status"] = df["Order_Status"].fillna("Unknown")

# Display cleaned dataset
print("Data cleaning completed!")
display(df.head())
```

```python
# STEP 6: VALIDATE CLEANED DATA

print("===== DATASET SHAPE =====")
print("Rows and Columns:", df.shape)

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

print("\n===== DUPLICATE ROWS =====")
print("Duplicate rows:", df.duplicated().sum())

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SAMPLE OF CLEANED DATA =====")
display(df.head())
```

```python
# STEP 7: HANDLE MISSING SALES AMOUNTS

# Calculate the median sales amount
median_sales = df["Sales_Amount"].median()

print("Median Sales Amount:", median_sales)

# Fill missing Sales_Amount values with the median
df["Sales_Amount"] = df["Sales_Amount"].fillna(median_sales)

print("\nMissing Sales Amounts after cleaning:",
      df["Sales_Amount"].isnull().sum())
```

```python
# STEP 8: FINAL VALIDATION

print("===== FINAL DATASET CHECK =====")

print("\nShape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFinal 5 Rows:")
display(df.tail())
```

```python
# STEP 9: SAVE COMPLETE CLEANED DATASET

# Save the cleaned dataset as a new CSV file
cleaned_file = "CodeOrbit_Task1_Cleaned_Dataset.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", cleaned_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
